```python
import numpy as np 
import pandas as pd 
from sklearn.model_selection import train_test_split
```

```python
df = pd.read_csv("Churn_Modelling.csv")
df.head() 
```

```python
df.info()
```

```python
df['Exited'].value_counts()
```

```python
df.drop(columns=['RowNumber', 'CustomerId', 'Surname'], errors='ignore', inplace=True)
```

```python
df = pd.get_dummies(df, columns=['Geography', 'Gender'], drop_first=True)
```

```python
df.head()
```

```python
X = df.drop(columns=['Exited'])
y = df['Exited'].values
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

X_train_trf = scaler.fit_transform(X_train)
X_test_trf = scaler.transform(X_test)
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential 
from tensorflow.keras.layers import Dense
```

```python
model = Sequential()
model.add(Dense(3,activation = 'sigmoid',input_dim = 11))
model.add(Dense(1,activation = 'sigmoid')) 
#this model contains 11 input nodes then 1 hidden layer with 3 nodes and 1 output layer with 1 node.
```

```python
model.summary()
```

```python
model.compile(loss='binary_crossentropy',optimizer = 'adam',metrics=['accuracy'])
model.fit(X_train_trf,y_train,epochs=100) 
```
